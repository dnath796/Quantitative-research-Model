# WTI forecast exploration
Run from the repository root or the notebooks directory after installing the package.
The notebook consumes the research library; results here are exploratory.

```python
from pathlib import Path
from quant_research import load_config, load_data, build_features, backtest, evaluate
root = Path.cwd() if Path("configs").exists() else Path.cwd().parent
config = load_config(root / "configs/demo.toml")
panel = load_data(config.data, [f.column for f in config.features])
dataset = build_features(panel, config.features, config.data.target)
panel.head()
```

```python
result = backtest(dataset, config.models, config.backtest)
metrics, comparisons = evaluate(result.forecasts, config.backtest.dm_lags)
metrics
```

```python
comparisons
```

```python
result.forecasts.pivot(index="target_date", columns="model", values="mean").plot(
    title="One-month-ahead predicted returns", figsize=(12, 4)
)
```

Change the config to `configs/wti.toml` for the existing workbook.
Predictor lags are availability assumptions; revised macro series are not vintage data.
Do not select specifications on this test sample and treat the resulting scores as untouched holdout evidence.
